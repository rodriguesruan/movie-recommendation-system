# MovieLens 1M - Data understanding

This notebook loads the raw MovieLens files and explores the data before any modeling step.


```python
from pathlib import Path
from src.data.loaders import load_movie_lens_data

raw_dir = Path('../data/raw/ml-1m')
ratings_df, movies_df, users_df = load_movie_lens_data(raw_dir)
print(f'Ratings: {len(ratings_df)}')
print(f'Movies: {len(movies_df)}')
print(f'Users: {len(users_df)}')
ratings_df.head()
```

```python
ratings_df['rating'].value_counts().sort_index().plot(kind='bar')
plt.title('Rating distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
user_counts = ratings_df.groupby('userId').size().sort_values(ascending=False)
movie_counts = ratings_df.groupby('movieId').size().sort_values(ascending=False)
user_counts.head(10)
movie_counts.head(10)
```

```python
genres = movies_df['genres'].str.split('|').explode()
genre_counts = genres.value_counts()
genre_counts.head(10).plot(kind='bar')
plt.title('Genre distribution')
plt.xlabel('Genre')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
